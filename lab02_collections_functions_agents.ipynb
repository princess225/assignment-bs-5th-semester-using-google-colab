# 7. Simple Rule-Based Agent


## Practice Questions

**Q1.** Write a rule-based agent that receives weather as sunny, rainy or snowy and returns an action.

```python
def weather_agent(weather):
    if weather == "sunny":
        return "Wear sunglasses"
    elif weather == "rainy":
        return "Take an umbrella"
    elif weather == "snowy":
        return "Wear a heavy jacket"
    return "Unknown weather"

for w in ["sunny", "rainy", "snowy"]:
    print(w, "->", weather_agent(w))
```

**Q2.** Write an agent that receives battery percentage and returns `Charge` when below 20, otherwise `Continue`.

```python
def battery_agent(percent):
    if percent < 20:
        return "Charge"
    return "Continue"

print(battery_agent(15))
print(battery_agent(65))
```

**Q3.** Write an agent that receives a traffic-light colour and returns the correct driving action.

```python
def driving_agent(color):
    if color == "red":
        return "Stop"
    elif color == "yellow":
        return "Slow down and wait"
    elif color == "green":
        return "Go"
    return "Unknown signal"

print(driving_agent("red"))
print(driving_agent("green"))
```

**Q4.** Write an agent that receives marks and attendance and returns `Eligible` or `Not Eligible`.

**Stated rule:** a student is *Eligible* if marks are at least 50 **and** attendance is at least 75%.

```python
def eligibility_agent(marks, attendance):
    if marks >= 50 and attendance >= 75:
        return "Eligible"
    return "Not Eligible"

print(eligibility_agent(70, 80))
print(eligibility_agent(70, 60))
print(eligibility_agent(40, 90))
```

**Q5.** Write an agent that receives `road_open` and `traffic_heavy` and chooses `Main Road` or `Alternate Road`.

```python
def road_agent(road_open, traffic_heavy):
    if road_open and not traffic_heavy:
        return "Main Road"
    return "Alternate Road"

print(road_agent(True, False))
print(road_agent(True, True))
print(road_agent(False, False))
```

**Q6.** Write an agent that receives a room status `clean` or `dirty` and returns `Move` or `Suck`.

```python
def room_agent(status):
    if status == "dirty":
        return "Suck"
    return "Move"

print(room_agent("dirty"))
print(room_agent("clean"))
```

**Q7.** Write an agent function that receives temperature and returns `Cold`, `Normal`, `Warm` or `Hot`.

```python
def temp_category_agent(temp):
    if temp < 15:
        return "Cold"
    elif temp < 25:
        return "Normal"
    elif temp < 32:
        return "Warm"
    return "Hot"

print(temp_category_agent(10))
```

**Q3.** Modify the agent so an unknown location returns `Unknown Location`.

```python
def vacuum_agent(location, status):
    if location not in ("A", "B"):
        return "Unknown Location"
    if status == "dirty":
        return "Suck"
    if location == "A":
        return "Right"
    return "Left"

print(vacuum_agent("C", "dirty"))
print(vacuum_agent("A", "clean"))
```

**Q4.** Store all chosen actions in a list and print the final list.

```python
chosen_actions = []
for location, status in four_percepts:
    chosen_actions.append(vacuum_agent(location, status))
print(chosen_actions)
```

**Q5.** Count how many times the action `Suck` occurs.

**Challenge 5.** Function `choose_cheapest(costs)` returning the smallest path cost from a list.

**Challenge 7.** Agent receiving location, weather and road status and recommending `Continue`, `Wait` or `Use Alternate Route`.

**Rules:** road blocked → *Use Alternate Route*; road open but weather bad (rainy/snowy/stormy) → *Wait*; otherwise → *Continue*.

```python
def travel_agent(location, weather, road_status):
    if road_status == "blocked":
        return "Use Alternate Route"
    if weather in ("rainy", "snowy", "stormy"):
        return "Wait"
    return "Continue"

print(travel_agent("Mingora", "sunny", "open"))
print(travel_agent("Madyan", "rainy", "open"))
print(travel_agent("Bahrain", "sunny", "blocked"))
```
