## Question 1
Create a list of six numbers and print the first three using slicing.

```python
numbers = [10, 20, 30, 40, 50, 60]
print("Full list:", numbers)
print("First three:", numbers[:3])
```

## Question 2
Create a list of five names and print the last three using slicing.

```python
names = ["Ali", "Sara", "Ahmed", "Zara", "Bilal"]
print("Full list:", names)
print("Last three:", names[-3:])
```

## Question 3
Create an empty list and add three KPK place names with `append()`.

```python
places = []
places.append("Peshawar")
places.append("Swat")
places.append("Abbottabad")
print("KPK places:", places)
```

## Question 4
Insert the value 99 at index 1 in a list.

```python
values = [5, 15, 25, 35]
print("Before:", values)
values.insert(1, 99)
print("After inserting 99 at index 1:", values)
```

## Question 5
Remove one specified value from a list.

```python
fruits = ["apple", "banana", "mango", "orange"]
print("Before:", fruits)
fruits.remove("mango")
print("After removing 'mango':", fruits)
```

## Question 6
Use `pop()` to remove the last item and print both the removed item and the remaining list.

```python
colors = ["red", "green", "blue", "yellow"]
removed_item = colors.pop()
print("Removed item:", removed_item)
print("Remaining list:", colors)
```

## Question 7
Sort a list of five path costs from smallest to largest.

```python
path_costs = [45, 12, 78, 23, 56]
print("Before sorting:", path_costs)
path_costs.sort()
print("After sorting (smallest to largest):", path_costs)
```

## Question 8
Reverse a list of four actions and print it.

```python
actions = ["move forward", "turn left", "turn right", "stop"]
print("Original:", actions)
actions.reverse()
print("Reversed:", actions)
```
