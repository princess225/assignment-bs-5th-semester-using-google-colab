## 12. if-elif-else

Use if-elif-else when there are more than two possible decisions. Conditions are checked from top to bottom.

```python
marks = int(input("Enter marks: "))
if marks >= 80:
    print("A")
elif marks >= 70:
    print("B")
elif marks >= 60:
    print("C")
elif marks >= 50:
    print("D")
else:
    print("F")
```

## 13. Nested if Statements

A nested if places one decision inside another. Use it only when the second decision depends on the first.

**AI connection:** Decision rules are a foundation of classical rule-based AI and expert systems.

**Example 1**

```python
age = int(input("Enter age: "))
has_id = input("Do you have an ID? ")
if age >= 18:
    if has_id == "yes":
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Age requirement not met")
```

**Example 2**

```python
marks = int(input("Enter marks: "))
attendance = int(input("Enter attendance: "))
if marks >= 50:
    if attendance >= 75:
        print("Course passed")
    else:
        print("Attendance requirement not met")
else:
    print("Marks requirement not met")
```

**Extra Example (Grading only if the student passed) — added for extra practice**

```python
marks = int(input("Enter marks: "))
if marks >= 50:
    if marks >= 90:
        print("Grade A+")
    else:
        print("Passed")
else:
    print("Failed")
```

## 14. while Loops

A while loop repeats code while a condition remains True. Make sure something changes so the loop can eventually stop.

**Why it matters:** AI programs often repeat operations over data, search states, training steps or user interactions.

**Example 1**

```python
count = 1
while count <= 5:
    print(count)
    count = count + 1
```

**Example 2**

```python
number = 5
while number >= 1:
    print(number)
    number = number - 1
```

**Example 3**

```python
password = ""
while password != "ai123":
    password = input("Enter password: ")
print("Access granted")
```

**Extra Example (Summing numbers from 1 to 10 with a while loop) — added for extra practice**

```python
n = 1
total = 0
while n <= 10:
    total = total + n
    n = n + 1
print("Sum 1 to 10:", total)
```

## 17. break and continue

break stops a loop immediately. continue skips the rest of the current iteration and moves to the next one.

**Why it matters:** AI programs often repeat operations over data, search states, training steps or user interactions.

**Example 1**

```python
for i in range(1, 11):
    if i == 6:
        break
    print(i)
```

**Example 2**

```python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

**Example 3**

```python
while True:
    command = input("Enter quit to stop: ")
    if command == "quit":
        break
    print("You entered", command)
```

**Extra Example (Printing odd numbers up to 15 only) — added for extra practice**

```python
for i in range(1, 21):
    if i % 2 == 0:
        continue
    if i > 15:
        break
    print(i)
```
